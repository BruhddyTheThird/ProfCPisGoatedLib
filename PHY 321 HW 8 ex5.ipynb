## Exercise 5 (35 pt) Motion inside a paraboloid

Let's figure out the motion of a particle stuck inside a paraboloid. The potential energy of the bead is given by $U = mgz$, where $m$ is the mass of the bead, $g$ is the acceleration due to gravity. The coordinates are $(x, y, z)$ in the Cartesian coordinate system.

- 5a. (5 pt) Write down the Lagrangian for the system in terms of $x$, $y$, and $z$ and their associated velocities $\mathcal{L}(x, y, z, \dot{x}, \dot{y}, \dot{z})$. Find the 3 equations of motion.

- 5b. (15 pt) The bead is constrained by the paraboloid $z = \frac{1}{2}k(x^2 + y^2)$. Find the constraint equation and use it to eliminate one of the coordinates from the Lagrangian. Find the 2 equations of motion. Here it can be useful to change to cylindrical coordinates $(\rho, \phi, z)$. Hint it should be that the Lagrangian is a function of $\rho$ and $\phi$ and their associated velocities, $\mathcal{L}(\rho, \dot{\rho}, \phi, \dot{\phi})$.

- 5c. (15 pt) For an initial condition where the bead is moving relative to the parabolic at some height above the bottom of the paraboloid, numerically solve the equations of motion (i.e., for $\rho(t)$ and $\phi(t)$) and plot the the motion of the bead. Does it stick to the paraboloid?

#5a Lagrangian in Terms of x, y, and z and Velocities

Kinetic Energy: T =(1/2)m(xdot^2 + ydot^2 + zdot^2)

Potential: U = mgz

Lagrangian: L = T - U = (1/2)m(xdot^2 + ydot^2 + zdot^2) - mgz

Equations of motion
x: d/dt(mxdot) = 0 -> xddot = 0

y: yddot = 0

z: d/dt(mzdot) = -mg -> zddot = -g


#5b Applying constraint z=1/2k(x^2+y^2)

Switch to cylindrical: x = pcos(phi) y = psin(phi)
z = 1/2 k (p^2)

xdot^2 + ydot^2 = pdot^2 + p^2phidot^2

zdot = kppdot
zdot^2 = k^2p^2pdot^2

Kinetic: T = 1/2 m (pdot^2 + p^2phidot^2 + k^2p^2pdot^2)
T = 1/2 m [(1+k^2p^2)pdot^2 + p^2phidot^2]

Potential: U = mgz = mg(1/2 k p^2) = 1/2mgkp^2

Lagrangian: T-U = 1/2m[(1+k^2p^2)pdot^2 + p^2phidot^2] - 1/2mgkp^2

Equations of motion
phi: d/dt(mp^2phidot) = 0 ---> Angular Momentum is conserved

rho: (1+k^2p^2)pddot + k^2p*pdot^2 - p*phidot^2 + gkp = 0




```python
#5c Numerically Solving for equatinos of motion and plotting
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

#constant values
m = 1.0
g = 9.81
k = 0.5

#ODEs
def equations(t,Y):
    rho, rhodot, phi, phidot = Y

    rhoddot = (rho*phidot**2 - g*k*rho - k**2*rho*rhodot**2) / (1+k**2*rho**2)

    phiddot = -2 * rhodot * phidot / rho
    return rhodot, rhoddot, phidot, phiddot

#Initial Conditions
rho0 = 1.0
rhodot0 = 0.0
phi0 = 0.0
phidot0 = 2.0

Y0 = [rho0, rhodot0, phi0, phidot0]

#Solve IVP
t_span = (0,10)
t_eval = np.linspace(0,10,1000)

sol = solve_ivp(equations, t_span, Y0, t_eval=t_eval)

rho = sol.y[0]
phi = sol.y[2]

#Convert to cartesian
x = rho * np.cos(phi)
y = rho * np.sin(phi)
z = 0.5 * k * rho**2

#Plot 3d
fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
ax.plot(x,y,z)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Bead Potential 3d Projection')
plt.show()
```

As we can see from the 3d projection, the "bead" does stick to the paraboloid as it travels up, down, and around the paraboloid.
