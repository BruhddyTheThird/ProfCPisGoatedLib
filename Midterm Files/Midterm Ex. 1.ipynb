## Exercise One
For the first exercise of the midterm, we will be working on the Magnus effect, which is governed by the equation,
$$\vec{F}_{magnus} = S(v) \vec{\omega}\times\vec{v},$$
Where $S(v)$ is a scalar function of speed, $\vec{\omega}$ is the angular velocity of the ball, and $\vec{v}$ is of course the velocity of the ball.
For reference, $S(v)$ will generally cause the ball to slow.

### 1a.)

We will model the stiuation of the drive of a golf ball off of a tee.
Specifically, we want to see the trajectory through 3-D space that a golf ball will take, when taking into account drag and magnus force. We also want to see how the height ($z$-axis) changes over time. We will then check our trajectories against the case of a smooth ball with no spin, and the case of a point particle with no air resistance, with similar initial conditions, as controls.

Our assumptions are as thus:
- No exterior wind force
- Cartesian co-ordinates
- Tee is at exactly (0,0,0)
- Air is a homogeneous, ideal gas.
- The ball will not experience forces from the ground.
- We will use average values for coefficients related to the characteristics of the balls.
- No effects on the balls due to humidity.
- We hit the ball towards positive $x$.
- The ball does not lose spin in flight. 

Since we are modeling strictly after impact, the forces on the ball are the force due to gravity, the drag force, and the magnus force.
$$ \vec{F}_g = -mg\hat{z} $$
$$ \vec{F}_d = -\frac{1}{2}C_d\rho A|\vec{v}|\vec{v}$$
$$ \vec{F}_M = S\vec{\omega}\times\vec{v}$$
We also find a value for $S$ based on a golf ball,
$$ S = \frac{2}{3}\rho r^3\pi = 2.5\cdot10^{-5}, $$
where $r=0.0213$.

### 1b.)

Before we dive into the differential equations, we state a few things about $\vec{\omega}$. This is a vector that exists for spinning objects, as the name "angular velocity" implies. For an object spinning in a given plane, (all 3D objects spin in exactly one unique plane) $\vec{\omega}$ is in the direction of the normal vector of that plane, with a pseudovector of length $\omega$ that is greater than zero for ccw (counterclock-wise) rotation and less than zero for cw (clock-wise) rotation. In the cartesian co-ordinate system, $\vec{\omega} = \omega_x \hat{x} +\omega_y \hat{y} +\omega_z \hat{z}$. In the case of a golf ball, as shown in the image below, $\omega_x$ will be zero, and most of the length of $\vec{\omega}$ will be contributed from $\omega_y$, as the majority of rotation is in the $x$-$z$ plane.

![Image explaining the above.](https://github.com/BruhddyTheThird/ProfCPisGoatedLib/blob/main/Midterm%20Files/Image%20I%20-%20PHY321%20Midterm%20I.png?raw=true)

Without doing too much of a deep dive into the mechanics of the strike of the golf ball, as we are more interested in the mechanics of the flight of the ball we will instantiate values for $\vec{\omega}$ instead of calculating them.
We find $\omega_y$ from averages of PGA golfers' swings (approx -330 rad/s).
We will also look at some examples where $\omega_z\neq0$, for which we take $\omega_z\in[-55,55]$ rad/s.

Our differential equations for the full case are as follows:
$$\ddot{\vec{r}} = -g\hat{z} - \frac{1}{2m}C_d\rho A|\dot{\vec{r}}|\dot{\vec{r}} + \frac{S}{m}\vec{\omega}\times\dot{\vec{r}}.$$
Which is a vector form. The cross product of angular and linear velocity is:
$$\vec{\omega}\times\vec{v} = (\omega_y v_z - \omega_z v_y)\hat{x} + \omega_zv_x\hat{y} -\omega_yv_x\hat{z}.$$
Thus, the scalar differential equations are:
$$\ddot{x} = -\frac{1}{2m}C_d\rho A\dot{x}\sqrt{\dot{x}^2 + \dot{y}^2 + \dot{z}^2} + \frac{S}{m}(\omega_y\dot{z}-\omega_z\dot{y})$$
$$\ddot{y} = -\frac{1}{2m}C_d\rho A\dot{y}\sqrt{\dot{x}^2+\dot{y}^2+\dot{z}^2}+\frac{S}{m}\omega_z\dot{x}$$
$$\ddot{z} = -g-\frac{1}{2m}C_d\rho A\dot{z}\sqrt{\dot{x}^2+\dot{y}^2+\dot{z}^2}-\frac{S}{m}\omega_y\dot{x}$$

Not only are these second order ODEs, but they are coupled ODEs. 
More specifically, it seems as though we need three differential equations to model the system, and double that if we are to model the system as a series of first-order differential equations.

### 1c.)

We can exchange them for first-order ODEs by the change of variables:
$$\begin{cases}
x=x, & v_x = \dot{x}\\
y=y & v_x = \dot{y}\\
z=z & v_z = \dot{z}
\end{cases}
$$
Which gives us that $\dot{x} = v_x,$ and $\ddot{x} = \dot{v}_x$, etc. for all other orthogonal cartesian directions in $\mathbb{R}^3$.

So we have:

$$ \dot{x} = v_x$$
$$ \dot{v}_x = -\frac{1}{2m}C_d\rho Av_x\sqrt{v_x^2+v_y^2+v_z^2}+\frac{S}{m}(\omega_yv_z-\omega_zv_y)$$
$$ \dot{y} = v_y $$
$$ \dot{v}_y = -\frac{1}{2m}C_d\rho Av_y\sqrt{v_x^2+v_y^2+v_z^2}+\frac{S}{m}\omega_zv_x$$
$$ \dot{z} = v_z $$
$$ \dot{v}_z = -g-\frac{1}{2m}C_d\rho Av_z\sqrt{v_x^2+v_y^2+v_z^2}-\frac{S}{m}\omega_yv_x$$

We can now write these as if they were in an Euler-Cromer numerical integrator.
Recalling the method itself, in one dimension, is is:

$$v_{n+1} = v_n + a_n\Delta t,$$
$$x_{n+1} = x_n+v_{n+1}\Delta t,$$
$$a_{n+1} = a(t+\Delta t).$$

To adapt this to our model we must realize that all orthogonal directions in $\mathbb{R}^3$ are linearly independent, so we only have to linearly scale the Euler-Cromer step.

$$ v_{x,n+1} = v_{x,n} + a_{x,n}\Delta t$$
$$ v_{y,n+1} = v_{y,n} + a_{y,n}\Delta t$$
$$ v_{z,n+1} = v_{z,n} + a_{z,n}\Delta t$$
$$ x_{n+1} = x_n + v_{x,n+1}\Delta t$$
$$ y_{n+1} = y_n + v_{y,n+1}\Delta t$$
$$ z_{n+1} = z_n + v_{z,n+1}\Delta t$$
$$ \dot{v}_{x,n+1} = -\frac{1}{2m}C_d\rho Av_{x,n+1}\sqrt{v_{x,n+1}^2+v_{y,n+1}^2+v_{z,n+1}^2}+\frac{S}{m}(\omega_yv_{z,n+1}-\omega_zv_{y,n+1})$$
$$ \dot{v}_{y,n+1} = -\frac{1}{2m}C_d\rho Av_{y,n+1}\sqrt{v_{x,n+1}^2+v_{y,n+1}^2+v_{y,n+1}^2}+\frac{S}{m}\omega_zv_{x,n+1}$$
$$ \dot{v}_{z,n+1} = -g-\frac{1}{2m}C_d\rho Av_{y,n+1}\sqrt{v_{x,n+1}^2+v_{y,n+1}^2+v_{y,n+1}^2}-\frac{S}{m}\omega_yv_{x,n+1}$$

And to re-iterate, the Euler-Cromer step is simply a restating of expanding the taylor series of $\vec{v}$ and $\vec{r}$ around $t+\delta t$.

### 1e.)

Of course, we have already talked about trends in the motion of the ball through discussion about the plots above, but something to re-iterate is that taking into account magnus effect, the maximum height and maximum downrange distance of the ball are noticeably greater than by only taking into account the drag on the ball.

We know that are values are at least close to correct, because we checked against the results from [similar numerical analyses](https://www.researchgate.net/publication/369468679_The_Influence_of_Magnus_Effect_on_The_Flying_Distance_of_Baseball_and_Golf?__cf_chl_rt_tk=RGvJzHEKbJf1JuBLE0urI3Spn7utKs5QAng5NryyI2M-1771624944-1.0.1.1-_tJzlY.ZqN7WmawQDGKacnx_I6oyxGzcTpLr03uA0Lo) from Zihang Ding, et. al, on the subject of trajectories of golf balls including the magnus effect and drag; our results are qualitatively close to their results. 


Something surprising is that our $z$ values in all cases are relatively small. Recalling initial research on the maximum height of drives in golf, our values are generally one-third of the value of real golf drives. In the same vein, something that implies that we are missing something in our modeling is the fact that empirically, most drives have a hang time of around 6-7 (funny number haha) seconds, while our trajectories indicate a hang time of only 4 seconds. 

Something to look into, which is discussed in the numerical analysis by Ding, et. al, is the influence of the Bernoulli effect on the trajectory of the ball, as the dimples on the ball should create a lift force on the ball, which should account for the error in empirical data to our numerical analysis.
