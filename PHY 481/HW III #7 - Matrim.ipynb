## Question VII - A Curious Distribution of Charge

We consider the integration of a function for which no anti-derivatives exist. 

Consider a line of charge that lives on the $x$-axis. It exists from $x=-1$ to $x=2$, and the distribution of the charge is given by the Gaussian,
$$\lambda(x) = 3e^{-x^2}$$

For the remainder of the problem, we aim to determine the total charge on this line by performing the integral,
$$Q = \int_{-1}^2 3e^{-x^2}\,\text{d}x$$

#### Part I

Plot the distribution of charge $\lambda(x)$ for $x\in[-1,2]$.

---

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
plt.style.use('ggplot')
%matplotlib inline
#imports
```

```python
#define function
def GaussLambda(x,bounds=(-200,200)):
    """
    x is a real number, bounds is a tuple.
    """
    if x<=xbounds[1] and x>=xbounds[0]:
        lam = 3*np.exp(-x**2)
    else:
        lam = 0
    return lam
```

```python
#initialize x
x = np.linspace(-3,3,1000)
xbounds=(-1,2)
#get y from function
y = np.array([GaussLambda(i,xbounds) for i in x])
```

#### Part III

Apply the trapezoidal rule to compute the integral of this function. Use 10 equal width steps.

---

We recall that the trapezoidal rule is,
$$I(a,b) \approx \frac{h}{2}(f(a)+f(b) + 2\sum_{k=1}^{N-1}f(x_k))$$
where, $N$ is the number of trapezoids, $h\equiv \frac{b-a}{N}$, and $x_k = a+kh$.

We can see that there are only really four parameters for the $I(a,b)$ function (which is our integral from $a$ to $b$):
- $a$ and $b$, the bounds of integration
- $f(x)$, the integrand function
- and $N$, the number of trapezoids to use (functionally, resolution).

We proceed to resolve these in the code.

```python
def TrapezoidIntegrate(func,bounds,N):
    a,b = bounds
    h = (b-a)/N # define height
    I=(h/2)*(func(a)+func(b)) #constant part of I
    k = 1
    while k <= N-1:
        x_k = a + k*h #Define x_k for each run of the loop
        I += h*(func(x_k)) #evaluate f(x_k) and add to constant part each run
        k+=1 #iterate
    return I
#define trapezoidal integration
    
```

```python
N=10
int_N = TrapezoidIntegrate(GaussLambda,xbounds,N=N)
print(f"Total charge is approximately {int_N:.6f} C")
```

#### Part V

Apply Simpson's rule to compute the integral of this function. Use 10 equal width steps.

---

We recall Simpson's rule,
$$I(a,b) \approx \frac{h}{3} (f(a) + f(b) + 4\sum_{k=1\equiv1\mod2}^{N-1}f(x_k)+2\sum_{k=2\equiv0\mod2}^{N-2}f(x_k))$$
where $N\equiv0\mod2$ is the number of parabolic segments, $h\equiv\frac{b-a}{N}$, and $x_k = a+kh$.

We see that the same parameters are used.

```python
#define function
def SimpsonsIntegrate(func,bounds,N):
    a,b=bounds
    h = (b-a)/N
    I = func(a)+func(b)
    k = 1
    while k <= N-1:
        x_k = a + k*h
        I += 2*(k % 2+1)*func(x_k) #2*(k%2+1) here is nice because it allows us to skip an if statement, which is more efficient.
        k+=1 #remember to increase k.
    I *= h/3
    return I
```

```python
N=10
int_N = SimpsonsIntegrate(GaussLambda,xbounds,N=N)
print(f"Total charge is approximately {int_N:.6f} C")
```

#### Part VIII

Explain in a few sentences how the trapezoidal rule and Simpson's rule are different, and how they compute the integrals. Explain how the Gaussian quadrature, id est, the method used by `quad` and its cousins in the `scipy.integrate` library, is different than both the trapezoidal and Simpson's rule.

---

The trapezoid rule splits the integral, very similarly to riemann integration, into a series of side-by-side quadrilaterals, titularly, trapezoids. The Simpson's rule, on the other hand, approximates, that f(x) is close to some quadratic function for every sub-interval. These approximations have errors on the order of $\frac{1}{N^2}$ and $\frac{1}{N^4}$, respectively, where $N$ is the number of sub-intervals. 

The method used by `quad`, Gaussian quadrature, uses an adaptive method to divide the function into very nice pieces, as opposed to dividing the function into equal pieces. Then, with nicer sections of the function, even linear fits can achieve better error minimization than the Simpson's rule. This is done through placing certain weights on parts of the function, and choosing irrational points as the endpoints of subsections.
