### Problem III - Plotting Vector Functions with `matplotlib`

We will plot us some vector fields, using `matplotlib.pyplot`.

I may use code snippets and methods from a library I created two years ago for the purpose of plotting and evaluating many sorts of mathematical constructs in calculus III. It is called MattyCLib, and it can be found on github [here](https://github.com/BruhddyTheThird/MattyCLib).

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
plt.style.use('ggplot')
%matplotlib inline
```

#### I:
$$\vec{v}(x,y) = \mathbf{r}\hat{r}$$

```python
def v(x,y):
    #r\hat{r} = x\hat{x} + y\hat{y}
    v_1 = x
    v_2 = y
    return np.meshgrid(v_1,v_2)
#arrow start points (i.e. our grid)
x,y = np.arange(-10,10,1),np.arange(-10,10,1)
#get our vectors
v_x,v_y = v(x,y)

fig=plt.figure()

plt.figure(figsize=(10,10))
plt.quiver(x,y,v_x,v_y,angles='xy',scale = 250)
plt.show()
```

#### II:
$$\vec{v}(x,y)=\frac{\hat{r}}{r}$$

Recall that $\hat{r} = \frac{\mathbf{r}}{|\mathbf{r}|}$, so
$$\hat{r}/r = \frac{\mathbf{r}}{x^2 + y^2} = \frac{x\hat{x}}{x^2+y^2} + \frac{y\hat{y}}{x^2+y^2}$$


```python
# similar:
def v(x,y):
    v_1 = x/(x**2+y**2) #I used ^ instead of **... which is the bitwise XOR operator, instead of square operator.
    v_2 = y/(x**2+y**2)
    return v_1,v_2

x_0,y_0 = np.arange(-10,10,0.5),np.arange(-10,10,0.5)
x, y = np.meshgrid(x_0,y_0)
v_x,v_y = v(x,y)

fig=plt.figure()

plt.figure(figsize=(10,10))
plt.quiver(x,y,v_x,v_y)
plt.xlim((-5,5))
plt.ylim((-5,5))
plt.show()
```

#### III:
$$\vec{v}(x,y) = \frac{x}{\sqrt{x^2+y^2}^3}\hat{x} +\frac{y}{\sqrt{x^2+y^2}^3}\hat{y}$$

```python
def v(x,y):
    v_1 = x / ((x**2+y**2)**(3/2))
    v_2 = y / ((x**2+y**2)**(3/2))
    return v_1,v_2

x_0,y_0 = np.arange(-10,10,1),np.arange(-10,10,1)
x, y = np.meshgrid(x_0,y_0)
v_x,v_y = v(x,y)

fig=plt.figure()

plt.figure(figsize=(10,10))
plt.quiver(x,y,v_x,v_y,scale = 7.5)
plt.xlim((-5,5))
plt.ylim((-5,5))
plt.show()
```

#### IV:
$$\vec{v}(x,y) = \hat{\phi}$$

```python
def v(x,y):
    phi = np.arctan2(x,y)
    v_1 = -np.cos(phi)
    v_2 = np.sin(phi)
    return v_1,v_2

x_0,y_0 = np.arange(-10,10,0.5),np.arange(-10,10,0.5)
x, y = np.meshgrid(x_0,y_0)
v_x,v_y = v(x,y)

fig=plt.figure()

plt.figure(figsize=(10,10))
plt.quiver(x,y,v_x,v_y,scale=50)
plt.xlim((-5,5))
plt.ylim((-5,5))
plt.show()
```
