**In order for x = (- $\alpha$/$\beta$)^(1/2) to exist, either $\alpha$ or $\beta$ has to be negative and the other positive.**

* 1c. (5 pt) Plot the phase space for the undriven and undamped case. What are the trajectories of the system? What is the behavior of the system? Does it match your expectations from the potential? *The figure above only starts this discussion.*

```python
def DO_undamped_undriven(X,V):
    
    dX = V
    dV = -alpha1 * X - beta1 * X**3
    
    return dX, dV

def generate_phase_space(x_lim, v_lim, grid_size):

    x = np.linspace(x_lim[0], x_lim[1], grid_size)
    v = np.linspace(v_lim[0], v_lim[1], grid_size)
    
    X, V = np.meshgrid(x, v)
    
    dX, dV = DO_undamped_undriven(X, V)
    
    return X, V, dX, dV

x_lim = (-5, 5)
v_lim = (-5, 5)
grid_size = 20
X, V, dX, dV = generate_phase_space(x_lim, v_lim, grid_size)
```

```python
fig, axs = plt.subplots(2, 1, figsize=(8, 10))  # Two rows, one column
    
axs[0].quiver(X, V, dX, dV, color='C0')
axs[0].set_title('Quiver Plot of the Undamped, Undriven Duffing Oscillator Phase Diagram (alpha=1,beta=1)')
axs[0].set_xlabel(r'$x$')
axs[0].set_ylabel(r'$\dot{x}$')

axs[1].streamplot(X, V, dX, dV, color='C1')
axs[1].set_title('Stream Plot of the Undamped, Undriven Duffing Oscillator Phase Diagram (alpha=1,beta=1)')
axs[1].set_xlabel(r'$x$')
axs[1].set_ylabel(r'$\dot{x}$');
plt.show()
```

Below is an example of the damped, driven 1D duffing oscillator. Use this as a template to do a few other types. As a part of this, you can generate different trajectories and do the following for them.

**Poincare section for $\alpha$ = -1, $\beta$ = 1, $\gamma$ = 0.3, $\omega$ = 1.2**

```python
#x = sol.y[0]
#v = sol.y[1]
#t = sol.t
def makeCrosspointsArray(x,v,c,transpose=False):
    crosspoints = []
    #c is the x value we slice for the poincare section.

    for i in range(len(x) - 1):
        if ((x[i]-c) * (x[i+1]-c)) <= 0:     
            # (x[i+1]-c) crosses x=c then (x[i+1]-c) is negative so the product is negative, same for (x[i]-c) crosses zero, and zero if either is zero.
            x_avg = (x[i]+x[i+1])/2
            v_avg = (v[i]+v[i+1])/2
            #avg our values to get a middle estimate instead of a left estimate for the actual crossing points.
            if transpose == False: #allow transposing for better viewing.
                crosspoints.append([x_avg, v_avg])
            else:
                crosspoints.append([v_avg,x_avg])

    crosspoints = np.array(crosspoints)
    return crosspoints
```

```python


def trajAndPoincarePlot(alpha=-1,beta=1,delta=0.2,gamma=0.3,omega=1.2,y0=[1.0,0],t_interval=(0,100),c=1,transpose=False):
    """`c` is crossing line in x, and `t_interval` is a tuple from initial to final time.
    `y0` is a list containing position and velocity, respectively.
    Doesn't return anything, just adds the plot to the cache. After calling with change of variables, just call plt.show()."""
    t_span = np.linspace(*t_interval, 10000)
    sol_new = solve_ivp(DO_d_d, t_interval, y0, args=(alpha, beta, delta, gamma, omega), t_eval=t_span)
    x = sol_new.y[0]
    v = sol_new.y[1]
    #do integration

    crosspoints = makeCrosspointsArray(x,v,c,transpose=transpose)
    #call function we just made, and use an x value that gets hit at multiple points.
    fig, ax = plt.subplots(1,2, figsize=(12,8))
    ax[0].plot(sol_new.t,x)
    ax[0].set_xlabel(r'$t$ (s)')
    ax[0].set_ylabel(r'$x$ (m)')
    ax[0].set_title('Trajectory of Oscillation; Position-Time')
    ax[0].grid()
    #Plot trajectory
    
    ax[1].scatter(crosspoints[:,0], crosspoints[:,1])  #proper indexing to get arrays of x and v 
    if transpose == False:
        ax[1].set_xlabel(r'Position, $x$')
        ax[1].set_ylabel(r'Velocity, $v$')
        ax[1].set_xlim((c-abs(c/20),c+abs(c/20)))
    else: #pay attention to the actual labels being set here.
        ax[1].set_ylabel(r'Position, $x$')
        ax[1].set_xlabel(r'Velocity, $v$')
        ax[1].set_ylim((c-abs(c/20),c+abs(c/20)))
    ax[1].set_title(f'Poincaré section at x = {c} (crossings)')
    ax[1].grid()
    #plot of poincare section.
    return  
#function to call for easy access. Just plug in numbers and get graphs out.
```
