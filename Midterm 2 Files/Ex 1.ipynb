## Exercise 1, The Duffing Oscillator (80pt)

Consider the following equation of motion for a damped and driven oscillator that is not necessarily linear. This is the [Duffing equation](https://en.wikipedia.org/wiki/Duffing_equation), which describes a damped driven simple harmonic oscillator with a cubic non-linearity. It's equation of motion is given by:

$$\ddot{x} + \delta \dot{x} + \alpha x + \beta x^3 = \gamma \cos(\omega t).$$

where $\alpha$, $\beta$, $\delta$, $\gamma$, and $\omega$ are constants.

Below is a figure showing the [strange attractor](https://en.wikipedia.org/wiki/Attractor#Strange_attractor) of the Duffing Oscillator over four periods. It is locally chaotic, but globally it is confined. This is a common feature of some non-linear systems.

![Strange attractor](https://upload.wikimedia.org/wikipedia/commons/f/fc/Duffing_oscillator_strange_attractor_with_color.gif)

We focus on this one dimensional case, but we will analyze it in parts to put together a full picture of the dynamics of the system. If you are looking for useful parameter choices and initial conditions, we suggest using those listed on [the Wikipedia page](https://en.wikipedia.org/wiki/Duffing_equation).

Recall that the potential for this system can give rise to different kinds of behavior depending on the parameters. The figure below shows the potential for different choices of $\alpha$ and $\beta$. **In order for x = (- $\alpha$/$\beta$)^(1/2) to exist, either $\alpha$ or $\beta$ has to be negative and the other positive.**

**If $\alpha$ > 0 then x=0 is a stable equilibrium point and x = +- (-$\alpha$/$\beta$)^(1/2) is an unstable equilibrium point.**

**If $\alpha$ < 0  then x=0 is an unstable equilibrium point and x = +-(-$\alpha$/$\beta$)^(1/2) is a stable equilibrium point.**


```python
alpha1 = 1
beta1 = 1
alpha2 = -1
beta2 = -1
m = 1
x = np.linspace(-2,2,1000)
U1 = m * alpha1 * x**2/2 + m * beta1 * x**4/4
U2 = m * alpha2 * x**2/2 + m * beta1 * x**4/4
U3 = m * alpha2 * x**2/2 + m * beta2 * x**4/4



plt.plot(x, U1, color='b', label='alpha=1, beta=1')
plt.plot(x, U2, color='r', label='alpha=-1, beta=1')
plt.plot(x, U3, color='purple', label='alpha=-1, beta=-1')
plt.xlabel('x')
plt.ylabel('U(x)')
plt.title('Potential of the Undamped, Undriven Duffing Oscillator')
plt.grid()
plt.legend()
plt.show()
```

**For $\alpha$ = 1, $\beta$ = 1, we see a stable equilibrium point at x=0 since it lies at the bottom of the potential well. There are no other equilibrium points in this case since both $\alpha$ and $\beta$ are positive, the other equilibrium points are imaginary.**

**For $\alpha$ = -1, $\beta$ =1, we see a unstable equilibrium point at x=0 since it lies at the top of the local potential well. There are two other stable equilibrium points (x = -1, 1) corresponding to x = +-(- $\alpha$/$\beta$)^(1/2) since $\alpha$ is positive and $\beta$ is negative.**

**For  $\alpha$ = -1, $\beta$ =1, we see x=0 is an unstable equilibrium point since it's at the top of the local potential well. No other equilibrium points exist.**

* 1c. (5 pt) Plot the phase space for the undriven and undamped case. What are the trajectories of the system? What is the behavior of the system? Does it match your expectations from the potential? *The figure above only starts this discussion.*

```python
def DO_undamped_undriven(X,V):
    
    dX = V
    dV = -alpha1 * X - beta1 * X**3
    
    return dX, dV

def generate_phase_space(x_lim, v_lim, grid_size):

    x = np.linspace(x_lim[0], x_lim[1], grid_size)
    v = np.linspace(v_lim[0], v_lim[1], grid_size)
    
    X, V = np.meshgrid(x, v)
    
    dX, dV = DO_undamped_undriven(X, V)
    
    return X, V, dX, dV

x_lim = (-5, 5)
v_lim = (-5, 5)
grid_size = 20
X, V, dX, dV = generate_phase_space(x_lim, v_lim, grid_size)
```

```python
fig, axs = plt.subplots(2, 1, figsize=(8, 10))  # Two rows, one column
    
axs[0].quiver(X, V, dX, dV, color='C0')
axs[0].set_title('Quiver Plot of the Undamped, Undriven Duffing Oscillator Phase Diagram (alpha=1,beta=1)')
axs[0].set_xlabel(r'$x$')
axs[0].set_ylabel(r'$\dot{x}$')

axs[1].streamplot(X, V, dX, dV, color='C1')
axs[1].set_title('Stream Plot of the Undamped, Undriven Duffing Oscillator Phase Diagram (alpha=1,beta=1)')
axs[1].set_xlabel(r'$x$')
axs[1].set_ylabel(r'$\dot{x}$');
```

**The phase diagram for an undamped, undriven duffing oscillator with $\alpha$ = 1, $\beta$ = 1 shows oscillations around x = 0 similar to the similar harmonic oscillator. An initial position and velocity of a system would oscillate and reach the same point. I expected this since the graph of the  potential with these parameters resembled the SHO potential and they both only have a stable equilibrium point of x = 0.**

```python
from scipy.integrate import solve_ivp

alpha5 = -2
beta5 = 3

def DO_ud_ud(t, y, alpha5, beta5):  #ud_ud represents undamped, undriven
    x, v = y
    return [v, -alpha5 * x - beta5 * x**3]

y0 = [0.5, 0.5]   #Initial conditions are chosen to be x_0=0.5m, v_0=0.5m/s
t = np.linspace(0, 10, 10000)
sol_ud_ud = solve_ivp(DO_ud_ud, (0,10), y0, args=(alpha5, beta5), t_eval=t)
x1 = sol_ud_ud.y[0]
v1 = sol_ud_ud.y[1]
plt.plot(sol_ud_ud.t, x1)
plt.xlabel('Time [s]')
plt.ylabel('Position [m]')
plt.title('Trajectory with $x_0$=0.5m, $v_0$=0.5m/s for alpha=-2, beta=3')
plt.grid()
plt.show()
```

**The particle starts at 0.5m with velocity 0.5m/s and we observe an oscillation around the point (2/3)^(1/2) = 0.816m when calculated using the formula of equilibrium points of the system. This is expected since this point must be a stable equilibrium point when observing the graph of a similar potential using the same parameters as well as the phase diagram.**
