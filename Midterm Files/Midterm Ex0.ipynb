### Planning Your Computational Essay (6 points)

We hear from students that often the hardest part of developing a computational essay is finding a question worth investigating. So we are trying to help you with that. You are not expected to write the computational essay for this midterm. Instead, you are expected to plan it. 

Your answers to this question do not commit you to completing this project on the exact idea you present here. You can change your mind later. This is a good exercise to get you thinking about what you might want to do.

For each of the questions below, you should write a paragraph or two (minimum 100 words, maximum 200 words per question) to answer the question. You may include equations, code, or figures in any of your answers.

* **1c. (1 pt)** Who is your team and how are you going to communicate regularly? What is the question(s) you want to investigate? What background information do you need to understand the question(s)? Where can you find that information?
* **1d. (1 pt)** What is the motivation for the question(s)? Why is it interesting to you? Why is it interesting to others? 
* **1e. (1.5 pts)** What is the computational aspect of the question(s)? What will you calculate? What will you simulate? What will you visualize? Include at least one plot, figure, or image that relates to your topic.
* **1f. (1.5 pts)** Outline the structure of the setup and analysis that you think you need in order to answer the question(s) you have. What are the key steps you would need to take? What are the key challenges you anticipate?
* **1g. (1 pt)** How will you know if you have answered the question(s)? What are the key results you would be looking for?

**Alternative Format:** If you prefer, you can write up a single document that answers parts (c) through (g) in a style you prefer. That document should be roughly 400-800 words long and may include equations, code, or figures.

**1a.** We have learned about a new educational technique known as the computational essay. This is an argument presented in writing with code and figures that consolidate your argument. First, you present your motivation or the story you're trying to tell according to Stephen Wolfram in text. Then you write the CodeText which introduces the code you're about to run. The input or the code is also written. Finally, the output of the code is shown to prove your story. It would also be good practice to include background information that your audience may not be aware of just so they understand the purpose behind your essay. These computational essays are usually written in Jupyter notebooks as they combine both code and text. According to Odden and Caballero's paper, Jupyter notebooks have great use for data scientists and professional physicists. What I also found interesting is how the computational essay's effectiveness was demonstrated. In Odden and Caballero's paper, students enjoyed the creative freedom as the students chose their own approaches to solving the problem they were given. This motivated students and the essay technique was unlike traditional methods of teaching in physics where solutions were either right or wrong. My question is how can the computational essay's effectiveness be quantified and compared to traditional methods? Perhaps we could have a study which controls for traditional methods and compares with computational essay's implementation in the classroom.
